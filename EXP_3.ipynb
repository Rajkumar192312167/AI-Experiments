```python
from itertools import permutations

for p in permutations('0123456789', 8):
    s,e,n,d,m,o,r,y = p
    if s!='0' and m!='0':
        send = int(s+e+n+d)
        more = int(m+o+r+e)
        money = int(m+o+n+e+y)
        if send+more == money:
            print(send, '+', more, '=', money)
            break
```
