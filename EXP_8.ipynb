```python
cost = [[0, 10, 15],
        [10, 0, 20],
        [15, 20, 0]]

path = [0, 1, 2, 0]
total = 0

for i in range(len(path)-1):
    total += cost[path[i]][path[i+1]]

print("Path:", path)
print("Total cost:", total)
```
