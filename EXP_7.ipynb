```python
graph = [[2, 3], [4, 5], [6], [], [], []]
visited = []

def dfs(n):
    if n not in visited:
        print(n, end=" ")
        visited.append(n)
        for i in graph[n-1]:
            dfs(i)

dfs(1)
```
