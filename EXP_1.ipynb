```python
from heapq import *
s=(1,2,3,4,0,6,7,5,8); g=(1,2,3,4,5,6,7,8,0)
q=[(0,s,[])]; v=set()
while q:
 _,s,p=heappop(q)
 if s in v: continue
 v.add(s); p=p+[s]
 if s==g:
  for x in p: print(x[:3],x[3:6],x[6:])
  break
 z=s.index(0)
 for n in [z-3,z+3,z-1,z+1]:
  if 0<=n<9 and abs(n%3-z%3)+abs(n//3-z//3)==1:
   t=list(s); t[z],t[n]=t[n],t[z]; t=tuple(t)
   if t not in v: heappush(q,(len(p)+sum(abs(i//3-g.index(t[i])//3)+abs(i%3-g.index(t[i])%3) for i in range(9) if t[i]),t,p))
```
