```python
s = [1,2,3,4,0,6,7,5,8]
goal = [1,2,3,4,5,6,7,8,0]

while s != goal:
    z = s.index(0)
    if z == 4:
        n = 7
    elif z == 7:
        n = 8
    else:
        break
    s[z], s[n] = s[n], s[z]
    print(s[:3], s[3:6], s[6:])
```
