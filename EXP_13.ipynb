```python
def minimax(d, a, b, maxp):
    if d == 3:
        return 3
    if maxp:
        v = -999
        for i in range(2):
            v = max(v, minimax(d+1, a, b, False))
            a = max(a, v)
            if b <= a:
                break
        return v
    else:
        v = 999
        for i in range(2):
            v = min(v, minimax(d+1, a, b, True))
            b = min(b, v)
            if b <= a:
                break
        return v

print("Best value:", minimax(0, -999, 999, True))
```
